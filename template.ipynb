```python
#@title Python package and dependencies
from qmm import *
import sympy as sp
import pandas as pd
from IPython.display import display, Markdown, Math
sp.init_printing(use_latex='mathjax', mat_symbol_style='bold')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)
```

## Model structure

```python
G = import_digraph("test/mesocosm.json")
display(Math(r'^{\circ}\mathbf{A} = ' + sp.latex(create_matrix(G, form='signed'))))
```

## Stability analysis

```python
display(Markdown(sign_stability(G).to_markdown(index=False)))
```

```python
display(Markdown(feedback_metrics(G).to_markdown(index=False)))
```

```python
display(Markdown(determinants_metrics(G).to_markdown(index=False)))
```

```python
display(Markdown(conditional_stability(G).to_markdown(index=False)))
```

```python
display(Markdown(simulation_stability(G, n_sim=10000).to_markdown(index=False)))
```

## Press perturbation analysis

```python
from IPython.display import display, Math
display(Math(r'\text{adj}(-^{\circ}\mathbf{A}) = ' + sp.latex(adjoint_matrix(G, form='signed'))))
```

```python
display(Math(r'\mathbf{T} = ' + sp.latex(absolute_feedback_matrix(G))))
```

```python
display(Math(r'\mathbf{W} = ' + sp.latex(weighted_predictions_matrix(G, as_abs=True).evalf(2))))
```

```python
display(Math(r'\Pr\left(\text{sign}[\text{adj}(-^\circ\mathbf{A})] = \text{sign}[\text{adj}(-^\#\mathbf{A})]\right) = '))
sign_determinacy_matrix(G, method='average', as_abs=True).evalf(2)
```

```python
display(Math(r'\Pr\left(\text{sgn}\left[-{^\#\mathbf{A}}^{-1}\right]\right) = '))
numerical_simulations(G, n_sim=10000, dist="uniform", as_abs=True).evalf(2)
```

## Qualitative predictions

```python
index, columns = get_nodes(G, 'state'), get_nodes(G, 'state')
display(Markdown("Qualitative predictions (prediction weights):"))
table_of_predictions(weighted_predictions_matrix(G), t1=0.5, t2=1, index=index, columns=columns)
```

```python
display(Markdown("Qualitative predictions (sign determinacy)"))
table_of_predictions(sign_determinacy_matrix(G), t1=0.8, t2=1, index=index, columns=columns)
```

```python
index, columns = get_nodes(G, 'state'), get_nodes(G, 'state')
display(Markdown("Qualitative predictions (numerical simulations):"))
table_of_predictions(numerical_simulations(G), t1=0.8, t2=1, index=index, columns=columns)
```
